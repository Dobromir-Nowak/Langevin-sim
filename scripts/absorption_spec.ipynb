```python
import numpy as np
import pandas as pd
from scipy.interpolate import PchipInterpolator
from pathlib import Path
import matplotlib.pyplot as plt

csv_path = Path("../data/eyespot_absorption_kreimer2009_digitized.csv")
abs_df = pd.read_csv(csv_path)

lam_abs = abs_df["wavelength_nm"].to_numpy()
abs = abs_df["relative_absorption"].to_numpy()

plt.plot(lam_abs, abs)
```

```python
import numpy as np
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt

csv_path = Path("../data/mcwhl5_spectrum_digitized_from_plot_1nm.csv")
spec = pd.read_csv(csv_path)

lam_data = spec["wavelength_nm"].to_numpy()
S_data = spec["normalized_intensity"].to_numpy()
```

```python
condition = lam_abs>=400
lam_abs_selected = lam_abs[condition]
abs_selected = abs[condition]

# effective spectrum
S_effective = abs_selected*S_data

plt.plot(lam_abs_selected, S_effective)
plt.show()


save_path = Path("../data/mcwhl5_effective.csv")

pd.DataFrame({
    "wavelength_nm": lam_abs_selected,
    "effective_intensity": S_effective,
}).to_csv(save_path, index=False)
```
